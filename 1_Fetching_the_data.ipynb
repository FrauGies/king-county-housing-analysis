# Retrieving the data from PostgreSQL

This notebook establishes the connection to PostgreSQL and exports the data to a CSV file that will be loaded at the start of the analysis notebook (`2_Housing_Analysis`).


## 1) Setup

```python
#import packages
import pandas as pd
import psycopg2
import os
from dotenv import load_dotenv
```


In order to create a connection to the PostgreSQL database, the following credentials are loaded from a local `.env` file: 

- host = the address of the machine the database is hosted on
- port = the virtual gate number through which communication will be allowed
- database = the name of the database
- user = the name of the user
- password = the password of the user

Please note that this file is excluded from version control via `.gitignore`.

```python
#load .env file and set credentials
load_dotenv()

DATABASE = os.getenv("DATABASE")
USER_DB = os.getenv("USER_DB")
PASSWORD = os.getenv("PASSWORD")
HOST = os.getenv("HOST")
PORT = os.getenv("PORT")
```

## 2) Connecting via psycopg2

```python
#open connection to database
conn = psycopg2.connect(
    database=DATABASE, user=USER_DB, password=PASSWORD, host=HOST, port=PORT
)
```

```python
# import the data into a pandas dataframe
query_string = """SELECT kchd.*, kchs.date, kchs.price
FROM eda.king_county_house_details kchd
LEFT JOIN eda.king_county_house_sales kchs
ON kchd.id = kchs.house_id"""
df_housesales = pd.read_sql(query_string, conn)

# close the connection
conn.close()

#display data
df_housesales.head()
```

```python
# export the data to a csv-file
df_housesales.to_csv("data/eda_housesales.csv", index=False)
```
